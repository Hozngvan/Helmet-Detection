### Yêu cầu : Tính các biểu thức lambda sau và cho biết phát biểu trong ngôn ngữ tự nhiên của chúng là gì?

##### $i)  \lambda P.(P@nam)@\lambda P.giỏi(P)$

- Thực hiện phép biến đổi $\alpha$ cho biến $\lambda P$ thành $\lambda R$ do sự trùng tên, kết quả thu được biểu thức sau:
$$\lambda R.(R@nam)@\lambda P.giỏi(P)$$
- Biểu thức có độ ưu tiên cao nhất là: ($\lambda R.(R@nam)@\lambda P.giỏi(P)$). Áp dụng phép biến đổi $\beta$ với biến $\lambda R$, kết quả thu được như sau:
$$ \lambda P.giỏi(P)@nam$$
- Biểu thức có độ ưu tiên cao nhất là: ($\lambda P.giỏi(P)@nam$). Áp dụng phép biến đổi $\beta P$ với biến lambda P, kết quả thu được như sau:
$$giỏi(nam)$$

$\Rightarrow$ Phát biểu ứng với biểu thức lambda trên là: Nam thì giỏi

##### $ii) (\lambda P. \lambda Q. \exists X((P@X) $ ^ $(Q@X))@\lambda P.{quyển\_sách}(P))@\lambda P.hay(P). $

- Do có sự trùng biến lambda P trong hai hàm, thực hiện phép biến đổi alpha cho $\lambda P. hay(P)$ thành $\lambda R. hay(R)$, ta thu được:
$$ (\lambda P. \lambda Q. \exists X((P@X) \land (Q@X))@\lambda P.{quyển\_sách}(P))@\lambda P.hay(P). $$
- Biểu thức có độ ưu tiên cao nhất là: $$\lambda P. \lambda Q. \exists X((P@X) ^ (Q@X))@\lambda P.{quyển\_sách}(P)$$. Áp dụng phép biến đổi $\beta$ với biến lambda P, kết quả thu được như sau:
$$ (\lambda Q. \exists X((\lambda P.quyển\_sách(P)@X)\land (Q@X)))@\lambda P.hay(P)$$
- Biểu thức có độ ưu tiên cao nhất là: $\lambda P.quyển\_sách(P)@X$, áp dụng phép biến đổi $\beta$ thu được:
$$ (\lambda Q. \exists X((quyển\_sách(X))\land (Q@X)))@\lambda P.hay(P)$$
- Biểu thức có độ ưu tiên cao nhất là $$ (\lambda Q. \exists X((quyển\_sách(X))\land (Q@X)))@\lambda P.hay(P)$$, áp dụng phép biến đổi $\beta$ thu được:
$$ \exists X(quyển\_sách(X)\land (\lambda P.hay(P)@X))$$
- Biểu thức có độ ưu tiên cao nhất là $\lambda P. hay(P)@X$, áp dụng phép biến đổi $\beta$ thu được:
$$ \exists X(quyển\_sách(X)\land (hay(X)))$$


$\Rightarrow$ Phát biểu ứng với biểu thức lambda trên là: Có một quyển sách hay

##### $iii) \lambda P.(P@toàn)@(\lambda P. \lambda X.(P@\lambda Y.trồng(X, Y))@(\lambda P. \lambda Q. \exists X((P@X)\land (Q@x))@\lambda P.cây\_chanh(P)))$

- Biểu thức có độ ưu tiên cao nhất là: 
$\\$ 
$$\lambda P. \lambda Q. \exists X((P@X)\land (Q@x))@\lambda P.cây\_chanh(P)$$

- Do có sự trùng biến lambda P giữa 2 hàm nên ta dùng phép biến đổi alpha cho $\lambda P.cây\_chanh(P)$ thành $\lambda R.cây\_chanh(R)$, thu được:
$$\lambda P. \lambda Q. \exists X((P@X)\land (Q@x))@\lambda R.cây\_chanh(R) (1)$$ 

- Dùng phép biến đổi beta cho phép biểu thức (1), ta thu được:
$$\lambda Q. \exists X(\lambda R.cây\_chanh(R)@X)\land (Q@x) (2)$$ 

- Tiếp tục áp dụng phép biến đổi beta cho biểu thức có độ ưu tiên cao nhất $\lambda R.cây\_chanh(R)@X$ thu được:
$$\lambda Q. \exists X(cây\_chanh(X))\land (Q@x) (3)$$ 

- Sau khi thu được biểu thức (1) và (3) từ các phép biến đổi, biểu thức ban đầu thành:
$$ \lambda P.(P@toàn)@(\lambda P.\lambda X.(P@λY.trồng(X, Y ))@\lambda Q.∃X(cây\_chanh(X)∧(Q@X))) (4)$$

- Biểu thức có độ ưu tiên cao nhất là: 
$$\lambda P.λX.(P@\lambda Y.trồng(X, Y ))@\lambda Q.∃X(cây\_chanh(X) ∧ (Q@X))$$

- Do có sự trùng tên biến lambda X trong hai hàm, áp dụng phương pháp alpha biến X thành Z cho hàm sau, ta thu được:
$$\lambda P.λX.(P@\lambda Y.trồng(X, Y ))@\lambda Q.∃Z(cây\_chanh(X) ∧ (Q@Z))$$

- Áp dụng biến đổi beta ta thu được:
$$\lambda X.(\lambda Q.∃Z(cây\_chanh(Z) ∧ (Q@Z))@\lambda Y.trồng(X, Y))$$

- Biểu thức có độ ưu tiên cao nhất là: $\lambda Q.∃Z(cây\_chanh(Z) ∧ (Q@Z))@\lambda Y.trồng(X, Y)$, thực hiện phép biến đổi beta thu được:
$$\lambda X.∃Z(cây\_chanh(Z) ∧ (\lambda Y.trồng(X, Y )@Z))$$

- Biểu thức có độ ưu tiên cao nhất là: $\lambda Y.trồng(X, Y)@Z$, biến đôie beta thu được:
$$ \lambda X.∃Z(cây\_chanh(Z) ∧ (trồng(X, Z))) (5)$$

- Thực hiện xong phép toán (5) trên, biểu thức (4) sẽ thành:
$$ \lambda P.(P@toàn)@(\lambda X.∃Z(cây\_chanh(Z) ∧ (trồng(X, Z))))$$

- Thực hiện phéo biến đổi beta tiếp, ta thu được:
$$ \lambda X.∃Z(cây\_chanh(Z) ∧ (trồng(X, Z)))@toàn $$

- Thực hiện phép biến đổi beta, ta thu được: 
$$∃Z(cây\_chanh(Z) ∧ (trồng(toàn, Z)))$$

$\Rightarrow$ Phát biểu ứng với biểu thức lambda trên là: Toàn trồng một cây chanh

##### iv) $ (\lambda P.\lambda Q. \forall X((P@X) \rightarrow (Q@X))@\lambda P.con\_mèo(P))@(\lambda P. \lambda X.(P@\lambda Y.sợ(X, Y))@(\lambda P. \lambda Q. \forall X((P@X) \rightarrow (Q@X))@\lambda P.con\_chó(P)))$

- Biểu thức có độ ưu tiên cao nhất là: $\lambda P. \lambda Q. \forall X((P@X) \rightarrow (Q@X))@\lambda P.con\_chó(P)$, do có sự trùng tên biến lambda P trong hai hàm nên ta thực hiện phép biến đổi alpha sao cho $\lambda P.con\_chó(P)$ thành $\lambda R.con\_chó(R)$, thu được:
$$\lambda P. \lambda Q. \forall X((P@X) \rightarrow (Q@X))@\lambda R.con\_chó(R) (1)$$
- Thực hiện phép biến đổi $\beta$ cho biến lambda P, ta thu được:
$$ \lambda Q. \forall X((\lambda R.con\_chó(R)@X) \rightarrow (Q@X)) (2)$$

- Biểu thức có độ ưu tiên cao nhất trong (2) là $\lambda R.con\_chó(R)@X$, thực hiện phép biến đổi $\beta$ cho biến lambda R, ta thu được:
$$ \lambda Q. \forall X(con\_chó(X) \rightarrow (Q@X)) (3)$$

- Sau khi thu được biểu thức (3), biểu thức ban đầu trở thành:
$$(\lambda P.\lambda Q. \forall X((P@X) \rightarrow (Q@X))@\lambda P.con\_mèo(P))@(\lambda P. \lambda X.(P@\lambda Y.sợ(X, Y))@(\lambda Q. \forall X(con\_chó(X) \rightarrow (Q@X)))) (3')$$

- Biểu thức có độ ưu tiên cao nhất là: $\lambda P. \lambda Q. \forall X((P@X) \rightarrow (Q@X)) @ \lambda P.con\_mèo(P)$, do có sự trùng tên biến lambda P trong hai hàm nên ta thực hiện phép biến đổi alpha sao cho $\lambda P. con\_mèo(P)$ thành $\lambda V. con\_mèo(V)$, thu được:

$$\lambda P. \lambda Q. \forall X((P@X) \rightarrow (Q@X)) @ \lambda V.con\_mèo(V)$$

- Thực hiện phép biến đổi $\beta$ cho biến lambda P, ta thu được:
$$\lambda Q. \forall X((\lambda V.con\_mèo(V)@X) \rightarrow (Q@X)) (4)$$

- Biểu thức có độ ưu tiên cao nhất trong (4) là: $\lambda V.con\_mèo(V)@X$, áp dụng phép biến đổi $\beta$ cho biến lmabda V, ta thu được:
$$\lambda Q. \forall X((con\_mèo(X)) \rightarrow (Q@X)) (5)$$

- Sau khi thu được biểu thức (5), biểu thức (3') trở thành:
$$(\lambda Q. \forall X(con\_mèo(X) \rightarrow (Q@X)))@(\lambda P. \lambda X.(P@\lambda Y.sợ(X, Y))@(\lambda Q. \forall X(con\_chó(X) \rightarrow (Q@X)))) (6)$$

- Biểu thức có độ ưu tiên cao nhất trong (6) là:
$$\lambda P. \lambda X.(P@\lambda Y.sợ(X, Y))@(\lambda Q. \forall X(con\_chó(X) \rightarrow (Q@X)))$$

- Do có sự trùng tên biến lambda X trong hai hàm nên ta thực hiện pháp biến đổi alpha sao cho biến X trong $\lambda Q. \forall X(con\_chó(X) \rightarrow (Q@X))$ thành biến Z, thu được:
$$\lambda P. \lambda X.(P@\lambda Y.sợ(X, Y))@(\lambda Q. \forall Z(con\_chó(Z) \rightarrow (Q@Z)))$$

- Thực hiện phép biến đổi beta cho biến lmabda P với biểu thức trên, ta thu được:
$$\lambda X.(\lambda Q. \forall Z(con\_chó(Z) \rightarrow (Q@Z))@\lambda Y.sợ(X, Y))$$

- Tiếp tục thực hiện phép biến đổi beta cho biến lambda Q, ta thu được:
$$\lambda X.(\forall Z(con\_chó(Z) \rightarrow (\lambda Y.sợ(X, Y)@Z)))$$

- Tiếp tục thực hiện phép biến đổi beta cho biến lambda Y, ta thu được:
$$\lambda X.(\forall Z(con\_chó(Z) \rightarrow sợ(X, Z))) (7)$$

- Sau khi thu được biểu thứ (7), biểu thức (6) trở thành:
$$(\lambda Q. \forall X(con\_mèo(X) \rightarrow (Q@X)))@(\lambda X.(\forall Z(con\_chó(Z) \rightarrow sợ(X, Z)))) $$

- Do có sự trùng tên biến lambda X trong hai hàm nên ta thực hiện phép biến đổi alpha sao cho biến X trong hàm sau thành M, thu được:
$$\lambda Q. \forall X(con\_mèo(X) \rightarrow (Q@X))@\lambda M.\forall Z(con\_chó(Z) \rightarrow sợ(M, Z)) $$

- Thưc hiện phép biến đổi beta cho biến lambda Q, ta thu được:
$$\forall X(con\_mèo(X) \rightarrow (\lambda M.\forall Z(con\_chó(Z) \rightarrow sợ(M, Z))@X)) $$

- Thực hiện phéo biến đổi beta cho biến lambda M, ta thu được:
$$\forall X(con\_mèo(X) \rightarrow (\forall Z(con\_chó(Z) \rightarrow sợ(X, Z)))) $$

$\Rightarrow$ Phát biểu ứng với biểu thức lambda trên là: Tất cả con mèo đều sợ tất cả con chó

##### v) $\lambda P.(P@nam)@((\lambda P.\lambda Q.\lambda X.((P@X) ∧ (Q@X))@\lambda P.giỏi(P))@\lambda P.siêng(P))$

- Biểu thức có độ ưu tiên cao nhất là:
$$ \lambda P.\lambda Q.\lambda X.((P@X) ∧ (Q@X))@\lambda P.giỏi(P) $$

- Do có sự trùng tên biến lambda P ở hai hàm, nên ta áp dụng phương pháp biến đổi alpha, đổi tên biến lambda P ở hàm sau thành Z, thu được:
$$ \lambda P.\lambda Q.\lambda X.((P@X) ∧ (Q@X))@\lambda Z.giỏi(Z) $$

- Áp dụng phương pháp biến đổi beta cho biến lambda P, ta thu được:
$$ \lambda Q.\lambda X.((\lambda Z.giỏi(Z)@X) ∧ (Q@X)) $$

- Tiếp tục thực hiện phương pháp biến đôir beta cho biến làm Z, thu được:
$$ \lambda Q.\lambda X.(giỏi(X) ∧ (Q@X)) (1)$$

-Sau khi thu được biểu thức (1), biểu thức ban đầu trở thành:
$$\lambda P.(P@nam)@(\lambda Q.\lambda X.(giỏi(X) ∧ (Q@X))@\lambda P.siêng(P)) (2)$$

-Biểu thức có độ ưu tiên cao nhất là: 
$$\lambda Q.\lambda X.(giỏi(X) ∧ (Q@X))@\lambda P.siêng(P)$$

-Thực hiện phép biến đổi beta cho biến lambda Q, ta thu được:
$$\lambda X.(giỏi(X) ∧ (\lambda P.siêng(P)@X))$$

- Tiếp tục thực hiện phép biến đổi beta cho biến lambda P, ta thu được:
$$\lambda X.(giỏi(X) ∧ siêng(X)) (3)$$

- Sau khi thu được biểu thức 3, biểu thức (1) trở thành:
$$\lambda P.(P@nam)@\lambda X.(giỏi(X) ∧ siêng(X)) $$

-Thực hiện phép biến đổi beta cho biến lambda P, ta thu được:
$$(\lambda X.(giỏi(X) ∧ siêng(X))@nam) $$

- Tiếp tục thương hiện phép biến đổi beta cho biến lambda X, thu được:
$$(giỏi(nam) ∧ siêng(nam)) $$

$\Rightarrow$ Phát biểu ứng với biểu thức lambda trên là: Nam vừa giỏi vừa siêng
